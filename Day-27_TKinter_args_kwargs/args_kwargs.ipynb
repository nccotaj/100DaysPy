### Unlimited arguments using *args
Unlimited postitional arguments

```python
#Using *args lets us iterate over the inputed argumetns, meaning we can have varying numbers of inputs to the fucntion
def sum(*args):
    total = 0
    print(args[0])  #can also access args by index, since its a tupple, are argumetns are postional
    for n in args:
        total += n
    return total
```

```python
print(sum(1,3,4,5))
```

### Unlimited kewword arguments using **kwargs

```python
#kwargs are a dictinary 
def calculate(n, **kwargs):
    # print(kwargs)   
    # for key, value in kwargs.items():
    #     print(key)
    #     print(value)
    n += kwargs["add"]        #using the keyword to ket the value
    n *= kwargs["multiply"] 
    print(n)
```

```python
calculate(2, add=3, multiply=5)
```
