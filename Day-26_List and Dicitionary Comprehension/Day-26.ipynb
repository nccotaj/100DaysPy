## Comprehension

Formula for list comprehension:
newlist = [new_item for item in list]

```python
numbers = [1, 2, 3]
new_number = [n + 1 for n in numbers]

print(new_number)
```

Works with strings: Ex - take a string and turn it into a list of strings

```python
name = "Nick"
letters_list = [letter for letter in name]
print(letters_list)
```

Doubling each number in a range

```python
range_list = [item * 2 for item in range(1,5)]
print(range_list)
```

Can also do conditional list comprehension:
newlist = [new_item for item in list if test]

```python
names = ["Nick", "Sam", "Bart", "Christopher","Reginald"]

short_names = [name for name in names if len(name) < 5]
print(short_names)
```

# Dicionary Comprehension

newdict = {new_key:new_value for item in list}

newdict = {new_key:new_value for (key,value) in dict.items()}

```python
import random
names = ["Nick", "Sam", "Bart", "Christopher","Reginald"]

student_score = {student:random.randint(1,100) for student in names}
print(student_score)



```

```python
passed_students = {student:score for (student, score) in student_score.items() if score > 60}
print(passed_students)
```

# Iterate Dataframe

```python
student_dict = {
    "student": ["Angela", "James", "Lily"],
    "score": [89, 98, 99]
}

#for looping through dictionary

for (key, value) in student_dict.items():
    print (value)


```

```python
import pandas

#making a Datafroma from student_dict
student_data_frame = pandas.DataFrame(student_dict)


#loop through data frame using iterrows. Outputs pandas sereies

#this prints angelas score
for (index, row) in student_data_frame.iterrows():
    if row.student == "Angela":
        print(row.score)
    
```
