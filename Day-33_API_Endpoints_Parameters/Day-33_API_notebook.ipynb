## API Notes

API - Application Programming Interface: set of commands, functions, protocols, and objects taht we can use to create software or interact with external systems

API Endpoint: The location - often a URL

```python
import requests
from datetime import datetime as dt

response = requests.get(url="http://api.open-notify.org/iss-now.json")
print(response)
```

The output of the above code will be a response code

1xx: Hold on
2xx: Successful
3xx: No permision
4xx: Something is wrong on your end (404 it doesn't exist)
5xx: Something wrong on server end

We can use the module to raise error depending on teh response code:

```python
response.raise_for_status()
```

Can also use .json to access the json data

```python
postition_data = response.json()["iss_position"] #Type in the key of what you what
print(postition_data)

long_data = response.json()["iss_position"]["longitude"]
print(long_data)
```

API Parameters: Can give an input when making the API request, to get different info back 

This is what a call with parameters looks like in the url: https://api.sunrise-sunset.org/json?lat=41.051922&lng=-73.539482

```python
MY_LAT = 41.051922
MY_LNG = -73.539482


parameters = {
    "lat": MY_LAT,
    "lng": MY_LNG,
    "formatted": 0   #1 means 12 hour time, 0 means 24 hour time
}


sun_response = requests.get(url="https://api.sunrise-sunset.org/json", params=parameters)
sun_response.raise_for_status()

sun_data = sun_response.json()["results"]
sunrise = sun_data["sunrise"]
sunset = sun_data["sunset"]

time_now = dt.now()

print(time_now.hour)
print(sunrise.split("T")[1].split(":")[0]) #prints just the hour 
```
