### Errors and exceptions

#### Try - Except

try: "Code the might cause and exception"

except: "Do this if there WAS and exception"

else: "Do this if there were no excpetions"

finally: "Do this no matter what happens"

NOTE: Use explicit and multiple except blocks to handle the different exceptions

```python
## Example ##

try:
    file = open("file2.txt") ##If file does not exist this should throw an error

    a_dict = {"key": "value"}
    print(a_dict["fake_key"]) ## Should throw a key error
    
except FileNotFoundError:                     #### Should have multiple except blocks to handle the different errors, although the first error takes precedence
    print ("No file, creating file now")
    file = open("file2.txt","w")
    file.write("something")
    
except KeyError as error_message:
    print(f"{error_message} not a valid key")

finally:
    file.close()

```

Raise exceptions. I can raise my own error

```python
height = 3.4
weight = 67

## Use raise to throw our own error
if height > 3:
    raise ValueError("Height should not be over 3 meters")


bmi = weight/height ** 2
print(bmi)
```
